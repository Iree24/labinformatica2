#### 1.**Max tra due numeri**
Scrivi un programma che chieda due numeri all'utente tramite la funzione input e mostri il più grande tra i due utilizzando la funzione print. Per quanto Python disponga di una funzione max(), siete invitati ad utilizzare le istruzioni istruzioni if, elif ed else per la scrittura dell'algoritmo.

```python
# 1. Acquisizione dell'input dall'utente e conversione in float
num1 = float(input("Inserisci il primo numero: ")) # se mettessi solo input() mi restituisce sempre una stringa, e tratta tipo '10'  e '2' come una parola del dizionario, confronta due stringhe e quindi 10<2 perchè prende 1<2 i caratteri iniziali
num2 = float(input("Inserisci il secondo numero: "))

# se avessi messo a = int(input("Inserisci il primo numero: ")) e b = int(input("Inserisci il secondo numero: "))
# python legge quello che l'utente digita come una stringa di testo (str) ovvero se digita 5 legge '5', e per fare dei confronti matematici devo convertire il testo in numero. con float mi legge sia con numeri interi che decimali

# 2. Algoritmo di confronto
if num1 > num2:
    print("Il numero più grande è:", num1) # primo è strettamente maggiore
elif num2 > num1:
    print("Il numero più grande è:", num2) # else if secondo strettamente maggiore
else:
    print("I due numeri sono uguali!")
```

#### **2.Max tra tre numeri**
Scrivi un programma che chieda tre numeri a, b, c all'utente e mostri il più grande tra loro.

```python
a = float(input("Inserisci il primo numero (a): "))
b = float(input("Inserisci il secondo numero (b): "))
c = float(input("inserisci il terzo numero (c): "))

#Metodo 1: operatore logico AND (a è il maggiore se è contemporaneamnte più grande sia di b che si c)
if a >= b and a >= c:
    print("Il numero più grande è:", a)
elif b >= a and b >= c:
    print("Il numero più grande è:", b)
else:
    print("Il numero più grande è:", c)

#Metodo 2: variabile di supporto

massimo = a #assumiamo che 'a' sia il massimo iniziale

if b > massimo:
    massimo = b
if c > massimo:
    massimo = c

print("Il numero più grande è:", massimo)
```

Il bug logico che si sarebbe creato: se avessimo usato solo >, con a = 5, b = 5 e c = 2 sarebbe successo questo deserto logico:
if a > b and a > c: 5 > 5 è False, quindi salta la prima condizione.
elif b > a and b > c: 5 > 5 è False, quindi salta anche la seconda!
Sarebbe finito nell'else, stampando come massimo c (cioè 2)! Un errore gravissimo!
Usando >=, invece, 5 >= 5 restituisce True e il programma individua correttamente 5 come massimo.

#### **3. Il maggiore tra tutti**
Scrivi un programma che, data una lista di numeri, fornisca in output il maggiore tra tutti gli elementi della lista.

```python
# 1. Definizione della lista di numeri
numeri = [14, 3, 89, 22, 5, 89, 41]

# 2. Inizializziamo il massimo con il primo elemento della lista (indice 0)
massimo = numeri[0]

# 3. Ciclo for per scorrere ciascun elemento della lista
for num in numeri:
    if num > massimo:
        massimo = num  # Aggiorniamo il massimo se troviamo un numero più grande

# 4. Stampa del risultato
print("La lista è:", numeri)
print("Il numero più grande presente nella lista è:", massimo)
```

Non posso fare if/elif/else o operatore AND. 

- L'Algoritmo del Massimo
è come una gara:
1. Prendi il primo elemento della lista e lo nomini temporaneamente "campione" (massimo = numeri[0]).
2. Scorri tutti gli altri elementi della lista uno alla volta con un ciclo for.
3. Ad ogni passo ti chiedi: "Questo numero è più grande del mio campione attuale?"
Se sì, il nuovo numero diventa il nuovo campione (massimo = numero).
Se no, vai avanti senza fare nulla.
4. Alla fine del ciclo, il valore rimasto dentro massimo sarà il più grande di tutta la lista!

Nota: python avrebbe anche la funzione nativa max(numeri) che fa tutto questo in una riga.
Nota2: devo inizializzare al primo elemento e non a zero perchè se avessi una lista di tutti negativi ovviamente il massimo sarebbe stato 0 anche se non appartiene alla lista, mentre nuemri[0] vuol dire il primo elemento.

```python
#1.Definisco una lista di numeri
numeri = [2,3,4]

#2.Creo uno spazio nella memoria, che è una variabile per salvare il prodotto. è l'accumulatore
prodotto = 1 # lo imposto che parte da 1 perchè sennò quando faccio il ciclo for prodotto*numero =0 

#3.Ciclo for: prendo ciascun numero presente nella lista e ad ogni passaggio lo moltiplico
for numero in numeri:
    prodotto = prodotto * numero #oppure scrivo prodotto *= numero

#4.Stampo il risultato finale
print("Il risultato finale è: " + str(prodotto)) 
```

#### **8.Generatore di istogrammi**
Scrivi una semplice funzione che, data una lista di numeri, fornisca in output un istogramma basato su questi numeri, usando asterischi per disegnarlo.

Data ad esempio la lista [3, 7, 9, 5], la funzione dovrà produrre questa sequenza:
***, *******, *********, *****

```python
#1. definizione della funzione
def disegna_istogramma(lista_numeri): #in python posso nmoltiplicare una stringa per un numero intero, sennò in altri linguaggi avrei dovuto fare un ciclo for
    for numero in lista_numeri:
        print("*" * numero)  # print("*" *3) Per Python significa: ripeti "*" per 3 volte -> "***", numero essendo nel ciclo for mi cambia ad ogni giro automaticamente

#non serve che inizializzo numero perchè se scrivo prima numero = 0 non faccio danni ma è inutile perchè python nel for crea la variabile, assegna automaticamente il numero estratto dalla lista e poi finisce il compito finita la lista (se metto a zero all'inizio semplicmente python soprascrive il primo valore nella lista)

#2. definzione della lista
lista_numeri = [3,7,9,5] # è meglio che io la metta dopo la definzione della funzione, tanto basta che sia prima della sua chiamata dove la funzione fa qualcosa di effettivo
#ordine: def funzioni, variabili delle funzioni, chiamata funzioni

# 3. Chiamata della funzione
disegna_istogramma(lista_numeri)
```

#### **9.Scriviamo la nostra versione di len()**
Scrivi una funzione che restituisca la lunghezza di una stringa o lista passata come parametro. In sostanza, seppur presente, provate a scrivere la nostra versione della funzione len!

```python
#1.definisco la funzione
def mia_len(sequenza):
    conteggio = 0 #è una variabile locale la inizializzo dentro la funzione
    for elemento in sequenza:
        conteggio += 1
    return conteggio #la funzione non mi deve stampare niente ma deve calcolare un valore e restituirlo a chi la chiama come fa len()

#2.variabii della funzione
sequenza = ["mela", "pera", "1", "3.14"]

#3.chiamata della funzione o con il print
mia_len(sequenza)
print("Lunghezza lista:", mia_len(sequenza))

#nota:
# 2. Prova dei 4 casi (stessa identica funzione per tutti!)

# Caso A: Lista mista
lista_mista = ["mela", "pera", 1, 3.14]
print("Lunghezza lista mista:", mia_len(lista_mista))  # Output: 4

# Caso B: Lista di soli numeri
lista_numeri = [10, 20, 30, 40, 50]
print("Lunghezza lista numeri:", mia_len(lista_numeri))  # Output: 5

# Caso C: Lista di sole parole
lista_parole = ["fiume", "lago", "mare"]
print("Lunghezza lista parole:", mia_len(lista_parole))  # Output: 3

# Caso D: Una singola stringa (parola)
parola = "Fisica"
print("Lunghezza stringa:", mia_len(parola))  # Output: 6
```

#### **10.A ciascuno il suo**
Scrivi una funzione che data in ingresso una lista A contenente n parole, restituisca in output una lista B di interi che rappresentano la lunghezza delle parole contenute in A.
Questo esercizio può essere risolto anche usando una list comprehension.

```python
#Metodo 1: ciclo for e .appen()
def calcola_lunghezze(A):
    B = [] #inizializzo a zero, ovvero una lista vuota, la crea osennò python non sa cosa è 
    for parola in A:
        lunghezza = len(parola)
        B.append(lunghezza) #anche B.appen(len(parola))
    return B

A = ["mela", "banana", "kiwi"]
print(calcola_lunghezze(A))  # Output: [4, 6, 4]

#Metodo 2: list comprehension B = [lunghezza_o_trasformazione for elemento in lista_originale]
B = [len(parola) for parola in A] #crea una lista B contenente len(parola) per ogni parola prensente in A

#oppure con la lista comprehension con funzione:
def calcola_lunghezze_comprehension(A):
    return [len(parola) for parola in A]

A = ["mela", "banana", "kiwi"]
print(calcola_lunghezze_comprehension(A))  # Output: [4, 6, 4]
```
